```python
import torch.nn as nn
from transformers import GPT2LMHeadModel, GPT2Tokenizer, Trainer, TrainingArguments
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
```

```python
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
from datasets import load_dataset

ds = load_dataset("sh0416/ag_news")
train_ds = ds["train"]
test_ds = ds["train"]
```

```python
def group(examples):
    result = {
        "labels": 
    }

train_ds = train_ds.map(
    group,
    batched=True
)
```

```python
tokenizer.encode(train_ds["title"][0] + "\n" + train_ds["description"][0]), train_ds["label"][0]
```

```python
num_classes = len(train_ds["label"].unique())
```

```python
model.lm_head = nn.Linear(768, num_classes, bias=False)
```

```python
training_args = TrainingArguments(
                  "gpt2_124M",
                  per_device_train_batch_size=8,
                  num_train_epochs=3,
                  learning_rate=1e-5,
                )
```

```python
trainer = Trainer(model, training_args, train_dataset=train_ds, eval_dataset=test_ds)

trainer.train()
```
