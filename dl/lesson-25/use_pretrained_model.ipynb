```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python
MODEL_NAME = "mardonbekhazratov/gpt2-trained-from-scratch"

model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

model.to(device)
```

```python
def generate_text(model, tokenizer, text, max_new_tokens=50, device=None):
    if device is None:
        device = "cuda" if torch.cuda.is_available() else "cpu"

    model.eval()
    model.to(device)

    inputs = tokenizer(text, return_tensors='pt').to(device)
    
    with torch.no_grad():
        out_ids = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=0.8,
            top_k=50,
            pad_token_id=tokenizer.eos_token_id
        )

    output = tokenizer.decode(out_ids[0], skip_special_tokens=True)

    return output
```

```python
generate_text(model, tokenizer, "salom", max_new_tokens=100)
```
